```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from webdriver_manager.chrome import ChromeDriverManager
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
import pandas as pd  # Import pandas for saving to Excel
import time

# Set up Chrome options
chrome_options = Options()
chrome_options.add_experimental_option('prefs', {
    "plugins.always_open_pdf_externally": True  # Ensure PDFs are handled externally
})

# Initialize WebDriver with ChromeDriverManager and Chrome options
driver = webdriver.Chrome(service=Service(ChromeDriverManager().install()), options=chrome_options)

# List to store all extracted links
extracted_links = []

# Function to extract links on the current page
def extract_links():
    try:
        # Wait until the links appear on the page
        WebDriverWait(driver, 20).until(EC.presence_of_all_elements_located((By.CSS_SELECTOR, 'div.item-title a')))
        
        # Find the links in the specified structure
        list_items = driver.find_elements(By.CSS_SELECTOR, 'div.item-title a')
        
        for item in list_items:
            link_url = item.get_attribute('href')  # Extract the link URL
            if link_url:  # Ensure the link is not None
                extracted_links.append(link_url)  # Append the link to the list
                print(f"Extracted link: {link_url}")  # Print each link as it's extracted

        print(f"Found {len(list_items)} links on this page.")
    except Exception as e:
        print(f"Error extracting links: {e}")

# Start at the initial page with a list of projects
driver.get("https://www.adb.org/projects/documents/doctype/Project252FProgram%20Administration%20Manual")

while True:
    # Extract links on the current page
    extract_links()

    try:
        # Wait for the "Next" button to be clickable and then click it
        next_button = WebDriverWait(driver, 10).until(
            EC.element_to_be_clickable((By.CSS_SELECTOR, 'li.pager_item.pager_item-next a[title="Go to next page"]'))  # Adjusted selector
        )
        next_button.click()  # Click the "Next" button
        time.sleep(2)  # Optional: Add a delay to allow the next page to load completely
    except Exception as e:
        print("No more pages to navigate or error occurred while clicking next:", e)
        break  # Exit the loop if the "Next" button is not found or an error occurs

# Save extracted links to an Excel file
df = pd.DataFrame(extracted_links, columns=["Links"])  # Create DataFrame with a column "Links"
df.to_excel("ADB_xxx.xlsx", index=False)  # Save to Excel without index

print(f"\nExtracted {len(extracted_links)} links in total and saved to 'ADB_links.xlsx'.")

# Quit the driver
driver.quit()
```

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from webdriver_manager.chrome import ChromeDriverManager
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
import pandas as pd
import time  # For sleep if needed

# Set up Chrome options
chrome_options = Options()
chrome_options.add_experimental_option('prefs', {
    "plugins.always_open_pdf_externally": True  # Ensure PDFs are handled externally
})

# Initialize WebDriver with ChromeDriverManager and Chrome options
driver = webdriver.Chrome(service=Service(ChromeDriverManager().install()), options=chrome_options)

# Set to store all extracted links (to avoid duplicates)
extracted_links = set()

# Function to extract links on the current page
def extract_links(page_number):
    try:
        print(f"Extracting links from page: {page_number}")
        
        # Wait until the links appear on the page
        WebDriverWait(driver, 20).until(EC.presence_of_all_elements_located((By.CSS_SELECTOR, 'div.item-title a')))
        
        # Find the links in the specified structure
        list_items = driver.find_elements(By.CSS_SELECTOR, 'div.item-title a')
        
        # Extract and store unique links
        for item in list_items:
            link_url = item.get_attribute('href')  # Extract the link URL
            if link_url:  # Ensure the link is not None
                extracted_links.add(link_url)  # Use a set to avoid duplicates
                print(f"Extracted link: {link_url}")

        print(f"Found {len(list_items)} links on this page.")
    except Exception as e:
        print(f"Error extracting links from page {page_number}: {e}")

# Loop through pages from 0 to 82
for page in range(83):  # 0 to 82 inclusive
    # Construct the URL for the current page
    url = f"https://www.adb.org/projects/documents/doctype/Project252FProgram%20Administration%20Manual?page={page}"
    
    # Navigate to the constructed URL
    driver.get(url)
    
    # Attempt to extract links
    for attempt in range(3):  # Retry up to 3 times if extraction fails
        extract_links(page)
        time.sleep(1)  # Optional sleep to allow time for dynamic content to load
        break  # Break after a successful attempt; otherwise, retry

# Save extracted links to an Excel file
df = pd.DataFrame(list(extracted_links), columns=["Links"])  # Convert set to DataFrame
df.to_excel("ADB_links3333.xlsx", index=False)  # Save to Excel without index

print(f"\nExtracted {len(extracted_links)} unique links in total and saved to 'ADB_links2.xlsx'.")

# Quit the driver
driver.quit()
```
