## 1) Get all the URLs for downloading 

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from webdriver_manager.chrome import ChromeDriverManager
import time

# Set up Chrome options (No download functionality needed)
chrome_options = Options()
chrome_options.add_experimental_option('prefs', {
    "plugins.always_open_pdf_externally": True  # Ensure PDFs are handled externally
})

# Initialize WebDriver with ChromeDriverManager and Chrome options
driver = webdriver.Chrome(service=Service(ChromeDriverManager().install()), options=chrome_options)

# List to store all extracted links
extracted_links = []

# Function to extract links on the current page
def extract_links():
    try:
        # Use the updated CSS selector to find the links in the specified structure
        list_items = driver.find_elements(By.CSS_SELECTOR, 'li.g h3.r a')
        
        for item in list_items:
            link_url = item.get_attribute('href')  # Extract the link URL
            if link_url:  # Ensure the link is not None
                extracted_links.append(link_url)  # Append the link to the list
                print(f"Extracted link: {link_url}")  # Print each link as it's extracted

        print(f"Found {len(list_items)} links on this page.")
    except Exception as e:
        print(f"Error extracting links: {e}")

# Loop through pages 1 to 93
for page in range(1, 5):
    main_url = f"https://www.ebrd.com/find?keywords=appraisal&content-ssbds=true&dates-all=true&search-type=search-exact&page={page}&order-by-date=false"  # Construct the URL for each page
    driver.get(main_url)
    time.sleep(5)  # Wait for the page to load

    # Wait for 20 seconds before extracting the links
    time.sleep(20)

    # Extract links on the current page
    extract_links()

# Print all extracted links
print(f"\nExtracted {len(extracted_links)} links in total:")
for link in extracted_links:
    print(link)

# Quit the driver
driver.quit()
```

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from webdriver_manager.chrome import ChromeDriverManager
import time
import os
import logging

# Setup download directory
download_dir = "/Users/adilqasin/Documents/WBG/Scrapping/European Bank/All Downloads"

# Set up Chrome options
chrome_options = Options()
chrome_options.add_experimental_option('prefs', {
    "download.default_directory": download_dir,
    "plugins.always_open_pdf_externally": True,
    "download.prompt_for_download": False,
    "download.directory_upgrade": True,
    "safebrowsing.enabled": True
})

# Initialize WebDriver
driver = webdriver.Chrome(service=Service(ChromeDriverManager().install()), options=chrome_options)

# Set up logging
logging.basicConfig(filename='download_errors.log', level=logging.ERROR)

# Function to check if the file is downloaded
def is_file_downloaded(file_name, download_dir):
    file_path = os.path.join(download_dir, file_name)
    return os.path.exists(file_path)

# Function to wait until the file is downloaded
def wait_for_download(file_name, download_dir, timeout=45):
    file_path = os.path.join(download_dir, file_name)
    start_time = time.time()
    
    while not os.path.exists(file_path):
        time.sleep(1)
        if time.time() - start_time > timeout:
            return False
    return True

# Function to download documents from extracted links
def download_documents(links):
    for link in links:
        try:
            print(f"Downloading document from: {link}")
            driver.get(link)  # Navigate to the link to trigger download
            
            # Extract the filename from the URL
            file_name = link.split('/')[-1]
            
            # Wait until the file appears in the download directory
            if wait_for_download(file_name, download_dir):
                print(f"Downloaded: {file_name}\n")
            else:
                print(f"Download timed out for: {file_name}\n")
        
        except Exception as e:
            logging.error(f"Error downloading {link}: {e}")
            print(f"Error downloading {link}: {e}")

# Start downloading documents
try:
    download_documents(extracted_links)
finally:
    driver.quit()
```
