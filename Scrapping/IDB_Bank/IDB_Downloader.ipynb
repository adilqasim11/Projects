```python
import os
import time
import pandas as pd
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from concurrent.futures import ThreadPoolExecutor, as_completed

# Define download directory
download_dir = "/Users/adilqasin/Documents/WBG/Scrapping/IDB/speedy"
if not os.path.exists(download_dir):
    os.makedirs(download_dir)

# Configure Chrome options for Selenium
chrome_options = Options()
chrome_options.add_experimental_option('prefs', {
    "download.default_directory": download_dir,
    "plugins.always_open_pdf_externally": True,
    "download.prompt_for_download": False,
    "download.directory_upgrade": True,
    "safebrowsing.enabled": True
})

# Initialize WebDriver
driver = webdriver.Chrome(service=Service(ChromeDriverManager().install()), options=chrome_options)

# Function to check if the download is complete
def is_download_complete(file_name):
    for _ in range(3):  # Retry up to 3 times
        files = os.listdir(download_dir)
        if file_name in files and not any(f.endswith('.crdownload') for f in files):
            return True
        time.sleep(4)  # Wait for 8 seconds between checks
    return False

# Function to sanitize file names
def sanitize_filename(filename):
    return "".join(c if c.isalnum() or c in (' ', '_') else '_' for c in filename)

# Function to download a single document
def download_document(file_url, file_name):
    print(f"Downloading: {file_name}")
    driver.get(file_url)
    if is_download_complete(file_name):
        print(f"Downloaded: {file_name}")
    else:
        return

# Function to download all "TC Document" and "Loan Proposal" links in the "Implementation Phase" section
def download_all_documents(base_url):
    try:
        driver.get(base_url)

        # Locate the "Implementation Phase" section by heading
        implementation_phase_section = WebDriverWait(driver, 10).until(
            EC.presence_of_element_located((By.XPATH, "//idb-accordion-panel[.//idb-heading[contains(text(), 'Implementation Phase')]]"))
        )

        # Find all document cards within the "Implementation Phase" section
        document_cards = implementation_phase_section.find_elements(By.XPATH, ".//idb-document-card")

        # Create a thread pool to download documents in parallel
        with ThreadPoolExecutor(max_workers=5) as executor:
            futures = []
            for card in document_cards:
                detail_text = card.find_element(By.XPATH, ".//div[@slot='detail']").text
                if detail_text in ["TC Document", "Loan Proposal"]:
                    raw_file_name = card.find_element(By.XPATH, ".//div[@slot='heading']").text
                    file_name = sanitize_filename(raw_file_name.replace(" ", "_"))
                    file_url = card.get_attribute("url")
                    futures.append(executor.submit(download_document, file_url, file_name))

            # Wait for all downloads to complete
            for future in as_completed(futures):
                future.result()  # To raise exceptions if any occurred

    except Exception as e:
        print(f"Error occurred: {e}")

extracted_links = ["https://www.iadb.org/en/project/EC-L1295",
                   "https://www.iadb.org/en/project/RG-T4220"]

for link in extracted_links:
    download_all_documents(link)

# Close the driver after all downloads are complete
driver.quit()
```

```python
import os
import time
import pandas as pd
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from webdriver_manager.chrome import ChromeDriverManager  # Import added

# Configure Chrome options for Selenium (headless mode for efficiency)
chrome_options = Options()
chrome_options.add_argument("--headless")
chrome_options.add_argument("--disable-gpu")
chrome_options.add_argument("--no-sandbox")

# Initialize WebDriver
driver = webdriver.Chrome(service=Service(ChromeDriverManager().install()), options=chrome_options)

# List to store extracted document data
document_data = []



# Function to extract Sector information
def extract_sector():
    try:
        # Locate the project table containing the sector information
        sector_element = WebDriverWait(driver, 10).until(
            EC.presence_of_element_located((By.XPATH, "//idb-project-table-row[p[@slot='stat-type'][contains(normalize-space(), 'Sector')]]/p[@slot='stat-data']"))
        )
        return sector_element.get_attribute("textContent").strip()
    except:
        return "Unknown"  # Default if sector is not found

# Function to extract document names with extensions and sector information
def extract_document_names(base_url):
    try:
        driver.get(base_url)

        # Extract Sector Information
        sector = extract_sector()

        # Locate the "Implementation Phase" section
        implementation_phase_section = WebDriverWait(driver, 10).until(
            EC.presence_of_element_located((By.XPATH, "//idb-accordion-panel[.//idb-heading[contains(normalize-space(), 'Implementation Phase')]]"))
        )

        # Find all document cards within the section
        document_cards = implementation_phase_section.find_elements(By.XPATH, ".//idb-document-card")

        # Extract document names
        for card in document_cards:
            detail_text = card.find_element(By.XPATH, ".//div[@slot='detail']").text.strip()
            if detail_text in ["TC Document", "Loan Proposal"]:
                # Extract document name properly
                raw_file_name = card.find_element(By.XPATH, ".//div[@slot='heading']").get_attribute("textContent").strip()
                
                # Extract file URL
                file_url = card.get_attribute("url")

                # Extract file extension from URL (if available)
                if file_url:
                    file_extension = os.path.splitext(file_url.split("/")[-1])[-1] or ".pdf"
                else:
                    file_extension = ".pdf"

                
                # Ensure we have a valid document name
                if not raw_file_name:
                    raw_file_name = file_url.split("/")[-1] if file_url else "Unknown_Document"

                # Remove everything after ".cfm" (including ".cfm?id=...")
                raw_file_name = raw_file_name.split(".cfm")[0]

                # Extract the correct file extension if it exists in the filename
                if "." in raw_file_name:
                    file_extension = os.path.splitext(raw_file_name)[-1]
                else:
                    file_extension = ".pdf"  # Default assumption

                # Ensure we are not adding duplicated file extensions
                if not raw_file_name.lower().endswith(file_extension.lower()):
                    raw_file_name += file_extension

                # Create a sanitized document name
                file_name = raw_file_name



                # Save extracted data
                document_data.append({
                    "Project URL": base_url,
                    "Document Name": file_name,
                    "Document Type": detail_text,
                    "Sector": sector
                })

                print(f"Extracted: {file_name} ({detail_text}) - Sector: {sector}")

    except Exception as e:
        print(f"Error occurred while processing {base_url}: {e}")
        import traceback
        traceback.print_exc()  # Log full error traceback

extracted_links = ["https://www.iadb.org/en/project/EC-L1295",
                   "https://www.iadb.org/en/project/RG-T4220"]

# Iterate over each link and extract document names + sector information
for link in extracted_links:
    print(f"Processing link: {link}")
    extract_document_names(link)

# Save extracted data to an Excel file
output_excel_path = "/Users/adilqasin/Documents/WBG/Scrapping/IDB/sector_details_8.xlsx"
output_df = pd.DataFrame(document_data)
output_df.to_excel(output_excel_path, index=False)

print(f"Extraction complete. Data saved to: {output_excel_path}")

# Close the driver
driver.quit()

```
